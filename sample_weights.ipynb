```python
import numpy as np
import scipy
import matplotlib.pyplot as plt
import tensorflow as tf
import silence_tensorflow
from sklearn.metrics import mean_squared_error

import methods
```

```python
rng = np.random.default_rng(44)
x = rng.normal(0,1,size=50_000)

n, bins, __ = plt.hist(x, np.arange(-5,5,.1), density=True)

alpha = .5
epsilon = .0001

hist = np.histogram(x, bins=100)
hist_dist = scipy.stats.rv_histogram(hist, density=True)

x_values = np.arange(-5,5,.001)
f = np.maximum(1 - alpha * hist_dist.pdf(x_values), epsilon)
f = f / np.mean(f)

plt.plot(x_values, f)
plt.ylim(0,None)
```

```python
np.random.seed(44)
y_pred = tf.constant(np.random.uniform(0,1,50))
y_true = tf.constant(np.random.uniform(0,1,50))

denseweight_dist = hist_dist.pdf(np.arange(0,101,1))

dw_loss = methods.DenseWeight_Loss(tf.constant(denseweight_dist, dtype="float32"))

print('My Loss : ', dw_loss(y_true, y_pred))
print(mean_squared_error(y_true, y_pred))
```
