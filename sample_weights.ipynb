```python
import numpy as np
import scipy
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng(44)
x = rng.normal(0,1,size=50_000)

n, bins, __ = plt.hist(x, np.arange(-5,5,.1), density=True)

alpha = .5
epsilon = .0001

hist = np.histogram(x, bins=100)
hist_dist = scipy.stats.rv_histogram(hist, density=True)

x_values = np.arange(-5,5,.001)
f = np.maximum(1 - alpha * hist_dist.pdf(x_values), epsilon)
f = f / np.mean(f)

plt.plot(x_values, f)
plt.ylim(0,None)
```
